```python
import re
import pandas as pd
import kagglehub
from collections import Counter


path = kagglehub.dataset_download("seriousran/appletwittersentimenttexts")
print("Path to dataset files:", path)

df = pd.read_csv(f"{path}/apple-twitter-sentiment-texts.csv")


text_data = " ".join(df['text'].dropna().astype(str))


hashtags = re.findall(r"#\w+", text_data)


total_hashtags = len(hashtags)
top_10 = Counter(hashtags).most_common(10)

print(f"Total number of hashtags: {total_hashtags}")
print("Top 10 Hashtags:")
for tag, freq in top_10:
    print(f"{tag}: {freq}")
```
